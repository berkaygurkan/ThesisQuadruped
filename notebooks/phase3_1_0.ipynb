```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import glob

def plot_comparison(log_dir):
    # Monitor CSV dosyalarını bul
    csv_files = glob.glob(os.path.join(log_dir, "*.monitor.csv"))
    
    plt.figure(figsize=(12, 6))
    
    for file in csv_files:
        # Varyasyon ismini dosya adından çek
        variant_name = os.path.basename(file).replace("_monitor.csv", "")
        
        # CSV oku (ilk 2 satır header olduğu için atla)
        try:
            df = pd.read_csv(file, skiprows=1)
            # Hareketli ortalama (Smoothing)
            window_size = 50
            rewards_smoothed = df['r'].rolling(window=window_size).mean()
            
            plt.plot(df.index, rewards_smoothed, label=variant_name, linewidth=2)
        except Exception as e:
            print(f"Hata okunamadı {file}: {e}")

    plt.title("Phase 3.1.0 Prototype Comparison: Adaptation to Varying Physics", fontsize=14)
    plt.xlabel("Episodes", fontsize=12)
    plt.ylabel("Reward (Moving Avg 50)", fontsize=12)
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# Çizdir
plot_comparison(trainer.base_log_dir)
```

```python
# Notebook Hücresi: Robust Generalization Test (Debug Modu Açık)
import os
import glob
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
import torch
from stable_baselines3 import PPO
from sb3_contrib import RecurrentPPO
from stable_baselines3.common.vec_env import DummyVecEnv, VecFrameStack
from src.envs.wrappers.random_dynamics import RandomDampingWrapper

# --- 1. GEREKLİ SINIFLAR ---
class Float32ObservationWrapper(gym.ObservationWrapper):
    def __init__(self, env):
        super().__init__(env)
        self.observation_space = gym.spaces.Box(
            low=env.observation_space.low,
            high=env.observation_space.high,
            shape=env.observation_space.shape,
            dtype=np.float32
        )
    def observation(self, observation):
        return np.array(observation, dtype=np.float32)

def get_latest_log_dir(base_path="../data/logs"):
    # Notebook path fix
    if not os.path.exists(base_path):
        base_path = "data/logs"
    
    # Klasörleri tarihe göre sırala
    all_subdirs = [d for d in glob.glob(os.path.join(base_path, "Phase3_Compare_*")) if os.path.isdir(d)]
    if not all_subdirs:
        print(f"❌ HATA: '{base_path}' altında hiç eğitim klasörü bulunamadı!")
        return None
    return max(all_subdirs, key=os.path.getmtime)

# --- 2. EVALUATION MOTORU (AKILLI DOSYA BULUCU) ---
def find_model_file(log_dir, variant_name):
    """Model dosyasını farklı isimlendirme ihtimallerine göre arar."""
    candidates = [
        f"final_model_{variant_name}.zip", # Checkpoint'li koddan gelen
        f"model_{variant_name}.zip",       # İlk versiyon koddan gelen
        f"{variant_name}.zip"              # Manuel kayıt
    ]
    
    for filename in candidates:
        full_path = os.path.join(log_dir, filename)
        if os.path.exists(full_path):
            return full_path
    return None

def evaluate_model(model_path, variant_name, damping_range, n_episodes=30): # Hız için 30 ep.
    print(f"   -> Loading: {os.path.basename(model_path)}")
    
    # Ortam Kurulumu (OOD - Ağır Sürtünme)
    def make_env():
        env = gym.make("Reacher-v5")
        env = Float32ObservationWrapper(env)
        env = RandomDampingWrapper(env, min_damping=damping_range[0], max_damping=damping_range[1])
        return env

    env = DummyVecEnv([make_env])
    if "FrameStack" in variant_name:
        env = VecFrameStack(env, n_stack=4)
    
    # Modeli Yükle
    device = "mps" if torch.backends.mps.is_available() else "cpu"
    try:
        if "LSTM" in variant_name:
            model = RecurrentPPO.load(model_path, env=env, device=device)
        else:
            model = PPO.load(model_path, env=env, device=device)
    except Exception as e:
        print(f"   ⚠️ Model yüklenirken hata: {e}")
        return []

    # Test
    rewards = []
    for _ in range(n_episodes):
        obs = env.reset()
        done = False
        total_reward = 0
        lstm_states = None
        episode_starts = np.ones((1,), dtype=bool)
        
        while not done:
            if "LSTM" in variant_name:
                action, lstm_states = model.predict(obs, state=lstm_states, episode_start=episode_starts, deterministic=True)
            else:
                action, _ = model.predict(obs, deterministic=True)
            
            obs, reward, done, info = env.step(action)
            total_reward += reward[0]
            episode_starts = done
            
        rewards.append(total_reward)
    
    env.close()
    return rewards

# --- 3. ANA AKIŞ ---
log_dir = get_latest_log_dir()

if log_dir:
    print(f"📂 Hedef Klasör: {log_dir}")
    print(f"📄 Klasördeki Dosyalar: {os.listdir(log_dir)}") # Dosyaları görelim!
    
    ood_damping = [3.0, 5.0] # EĞİTİMDE GÖRÜLMEMİŞ ZORLUK
    target_models = ["model_Vanilla_PPO", "model_PPO_FrameStack", "model_LSTM"]
    results = {}

    print(f"\n🧪 BAŞLIYOR: OOD Testi (Damping {ood_damping})")
    for model_name in target_models:
        model_file = find_model_file(log_dir, model_name)
        
        if model_file:
            scores = evaluate_model(model_file, model_name, ood_damping)
            if scores:
                results[model_name] = scores
        else:
            print(f"❌ BULUNAMADI: {model_name} için model dosyası yok.")

    # --- 4. GRAFİK ÇİZİMİ (MATPLOTLIB CRASH FIX) ---
    if len(results) > 0:
        plt.figure(figsize=(10, 6))
        plt.title(f"Generalization Gap: Adaptation to Heavy Friction {ood_damping}", fontsize=14)
        
        # Boxplot verisi hazırlama
        labels = list(results.keys())
        data = list(results.values())
        
        plt.boxplot(data, labels=labels, patch_artist=True)
        plt.ylabel("Total Reward")
        plt.grid(True, alpha=0.3)
        
        # Tablo Raporu
        print("\n--- 📊 SONUÇ TABLOSU ---")
        print(f"{'Model':<15} | {'Mean Reward':<15} | {'Std Dev':<15}")
        print("-" * 50)
        for name, scores in results.items():
            print(f"{name:<15} | {np.mean(scores):.2f}          | {np.std(scores):.2f}")
            
        plt.show()
    else:
        print("\n⚠️ HİÇBİR MODEL TEST EDİLEMEDİ. Lütfen 'Klasördeki Dosyalar' çıktısını kontrol edin.")
```

```python
# Notebook Hücresi: Sudden Fault Tolerance Test (Non-Stationarity)
import os
import glob
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
import torch
from stable_baselines3 import PPO
from sb3_contrib import RecurrentPPO
from stable_baselines3.common.vec_env import DummyVecEnv, VecFrameStack

# --- 1. ÖZEL WRAPPER: ANİ ARIZA ENJEKTÖRÜ ---
class SuddenFaultWrapper(gym.Wrapper):
    """
    Belirli bir adımda (fault_step) fiziksel parametreleri aniden değiştirir.
    Senaryo: Damping 1.0 -> 5.0
    """
    def __init__(self, env, fault_step=40, normal_damping=1.0, fault_damping=5.0):
        super().__init__(env)
        self.fault_step = fault_step
        self.normal_d = normal_damping
        self.fault_d = fault_damping
        self.step_count = 0
        
    def reset(self, seed=None, options=None):
        self.step_count = 0
        # 1. Başlangıçta Normal Damping (1.0)
        if hasattr(self.env.unwrapped, 'model'):
            self.env.unwrapped.model.dof_damping[:] = self.normal_d
        return super().reset(seed=seed, options=options)

    def step(self, action):
        self.step_count += 1
        
        # 2. Arıza Anı Kontrolü
        if self.step_count == self.fault_step:
            if hasattr(self.env.unwrapped, 'model'):
                self.env.unwrapped.model.dof_damping[:] = self.fault_d
            # Bilimsel analiz için konsola değil, loga yazmak daha iyidir ama burada görelim
            # print(f"⚠️ [Step {self.step_count}] ANİ ARIZA: Damping {self.normal_d} -> {self.fault_d}")

        return super().step(action)

# --- 2. YARDIMCI SINIFLAR ---
class Float32ObservationWrapper(gym.ObservationWrapper):
    """Mac MPS uyumluluğu için."""
    def __init__(self, env):
        super().__init__(env)
        self.observation_space = gym.spaces.Box(
            low=env.observation_space.low,
            high=env.observation_space.high,
            shape=env.observation_space.shape,
            dtype=np.float32
        )
    def observation(self, observation):
        return np.array(observation, dtype=np.float32)

def get_latest_log_dir(base_path="../data/logs"):
    if not os.path.exists(base_path): base_path = "data/logs"
    all_subdirs = [d for d in glob.glob(os.path.join(base_path, "Phase3_Compare_*")) if os.path.isdir(d)]
    return max(all_subdirs, key=os.path.getmtime) if all_subdirs else None

def find_model_file(log_dir, variant_name):
    candidates = [f"final_model_{variant_name}.zip", f"model_{variant_name}.zip", f"{variant_name}.zip"]
    for f in candidates:
        path = os.path.join(log_dir, f)
        if os.path.exists(path): return path
    return None

# --- 3. ANALİZ MOTORU (STEP-BY-STEP REWARD) ---
def run_fault_test(model_path, variant_name, n_episodes=10):
    print(f"   -> Testing {variant_name}...")
    
    # Ortam: 100 Adımlık Uzun Epizot (Recovery görmek için)
    def make_env():
        # Gymnasium'da max_episode_steps parametresi ile süreyi uzatıyoruz
        env = gym.make("Reacher-v5", max_episode_steps=100)
        env = Float32ObservationWrapper(env)
        # Ani Arıza Enjekte Et (40. adımda)
        env = SuddenFaultWrapper(env, fault_step=10, normal_damping=1.0, fault_damping=5.0)
        return env

    env = DummyVecEnv([make_env])
    if "FrameStack" in variant_name:
        env = VecFrameStack(env, n_stack=4)

    device = "mps" if torch.backends.mps.is_available() else "cpu"
    try:
        if "LSTM" in variant_name:
            model = RecurrentPPO.load(model_path, env=env, device=device)
        else:
            model = PPO.load(model_path, env=env, device=device)
    except Exception as e:
        print(f"   ⚠️ Hata: {e}")
        return None

    # Veri Toplama: Her adımın ödülünü sakla
    # Shape: [Episode, Step]
    step_rewards = [] 
    
    for _ in range(n_episodes):
        obs = env.reset()
        done = False
        lstm_states = None
        episode_starts = np.ones((1,), dtype=bool)
        episode_rew = []
        
        while not done:
            if "LSTM" in variant_name:
                action, lstm_states = model.predict(obs, state=lstm_states, episode_start=episode_starts, deterministic=True)
            else:
                action, _ = model.predict(obs, deterministic=True)
            
            obs, reward, done, info = env.step(action)
            episode_rew.append(reward[0])
            episode_starts = done
            
            # 100. adımda bitmesi lazım ama VecEnv bazen otomatik reset atar,
            # biz manuel olarak 100 veriyi garanti edelim.
            if len(episode_rew) >= 100:
                break
        
        # Veriyi sabitle (bazen 99 veya 101 olabilir, kırp/doldur)
        if len(episode_rew) < 100:
            episode_rew += [episode_rew[-1]] * (100 - len(episode_rew))
        step_rewards.append(episode_rew[:100])
    
    env.close()
    return np.mean(step_rewards, axis=0) # Epizotların ortalamasını döndür

# --- 4. GÖRSELLEŞTİRME ---
log_dir = get_latest_log_dir()
if log_dir:
    print(f"📂 Analiz: {log_dir}")
    models = ["Vanilla_PPO", "PPO_FrameStack", "LSTM"]
    data = {}

    for m in models:
        path = find_model_file(log_dir, m)
        if path:
            avg_curve = run_fault_test(path, m)
            if avg_curve is not None:
                data[m] = avg_curve

    if data:
        plt.figure(figsize=(12, 6))
        
        # Arıza Çizgisi
        plt.axvline(x=10, color='r', linestyle='--', label='Sudden Fault (t=40)')
        plt.text(11, min([min(d) for d in data.values()]), 'Damping x5', color='red')

        # Performans Eğrileri
        for name, curve in data.items():
            linewidth = 3 if "LSTM" in name else 2
            alpha = 1.0 if "LSTM" in name else 0.6
            plt.plot(curve, label=name, linewidth=linewidth, alpha=alpha)

        plt.title("Online Adaptation Test: Sudden Physical Fault Recovery", fontsize=14)
        plt.xlabel("Simulation Steps (t)")
        plt.ylabel("Instantaneous Reward")
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.show()
        
        print("✅ Analiz Tamamlandı. Grafiği inceleyin.")
        print("LSTM eğrisi, kırmızı çizgiden sonra diğerlerine göre daha çabuk toparlanıyorsa hipotez doğrulanmıştır.")
    else:
        print("❌ Hiçbir model verisi toplanamadı.")
```
